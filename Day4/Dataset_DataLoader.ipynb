# Import Libraries

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

# Load Data

```python
df = pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
df.head()
```

```python
df.shape
```

```python
df.drop(columns={"id", "Unnamed: 32"}, inplace=True)
```

```python
df.head()
```

```python
X = df.iloc[:,1:]
```

```python
y = df.iloc[:,0]
```

# Split Data

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
```

### Feature Scaling

```python
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.fit_transform(X_test)
```

```python
X_train
```

### Label Encoding

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.fit_transform(y_test)
```

```python
y_train
```

```python
print(type(X_train))
print(type(y_train))
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor.dtype
```

```python
X_train_tensor = X_train_tensor.to(torch.float32)
X_test_tensor = X_test_tensor.to(torch.float32)
y_train_tensor = y_train_tensor.to(torch.float32)
y_test_tensor = y_test_tensor.to(torch.float32)
```

### Why `float64` → `float32`?
* PyTorch's neural network layers (`nn.Linear`) use `float32` by default, while Pandas/NumPy often provide `float64`.
* Converting to `float32` avoids dtype mismatch errors and is faster and more memory-efficient.

# Defining Model by PyTorch nn Module

```python
class MySimpleNN(nn.Module):

    def __init__(self,num_features):

      super().__init__() # call parent constructor
      self.linear = nn.Linear(num_features, 1)
      self.sigmoid = nn.Sigmoid()

    def forward(self, features):
      out = self.linear(features)
      out = self.sigmoid(out)

      return out


# create model
model = MySimpleNN(X_train_tensor.shape[1])
```

# Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

# Define the Loss Function and Optimizer

```python
# define loss function
loss_function = nn.BCELoss()

# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
```

# Trainig PipeLine

```python
batch_size = 45
n_samples = len(X_train_tensor)
```

```python
# Step - 1 define loop for epoch
for epoch in range(epochs):

    # Step - 2 Manual Mini-Batch
    for start_idx in range(0,n_samples,batch_size):
        end_idx = start_idx + batch_size
        X_batch = X_train_tensor[start_idx:end_idx]
        y_batch = y_train_tensor[start_idx:end_idx]

        # Step - 3 forward pass
        y_pred = model(X_batch)

        # Step - 4 loss calculate
        loss = loss_function(y_pred,y_batch.reshape(-1, 1))

        # Step - 5 Zero gradients
        optimizer.zero_grad()

        # Step - 6 backward pass
        loss.backward()

        # Step - 7 parameter update
        optimizer.step()

    print(f"epoch: {epoch + 1}, loss: {loss}")
```

### Problems with Manual Data Handling

1. **No Standard Interface**  
   Data is accessed manually using tensors and indexes. `Dataset` provides a standard way to access samples using `__len__()` and `__getitem__()`.

2. **No Easy Transformations**  
   Preprocessing and transformations have to be handled manually. `Dataset` allows transformations to be applied automatically when data is loaded.

3. **Shuffling & Sampling**  
   Data is processed in the same order unless we manually shuffle it. `DataLoader` provides `shuffle=True` and sampling support.

4. **Batch Management & Parallelization**  
   We manually create batches using `start_idx` and `end_idx`. `DataLoader` automatically creates batches and can load data in parallel using `num_workers`.

### In Short
`Dataset` → manages and preprocesses individual data samples.  
`DataLoader` → handles batching, shuffling, sampling, and parallel data loading.

# Dataset & DataLoader

Dataset and DataLoader are core abstractions in PyTorch that decouple how you define your data from how you efficiently iterate over it in training loops.

```python
from torch.utils.data import Dataset, DataLoader
```

Create custome dataset class

```python
class CustomDataset(Dataset):

    def __init__(self, features, labels):

        self.features = features
        self.labels = labels

    def __len__(self):

        return self.features.shape[0]

    def __getitem__(self, index):

        return self.features[index], self.labels[index]
```

create object of custom dataset class

```python
dataset = CustomDataset(X_train_tensor, y_train_tensor)
```

create object of DataLoader class

```python
dataloader = DataLoader(dataset, batch_size=45, shuffle=True)
```
