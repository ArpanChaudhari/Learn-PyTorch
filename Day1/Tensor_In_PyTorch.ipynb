```
import time
# Define the size of the matrices
size = 10000

# create random matrices on cpu

cpu1 = torch.randn(size , size)
cpu2 = torch.randn(size,size)

# measure time on cpu
start_time = time.time()
result_cpu = torch.matmul(cpu1,cpu2)
cpu_time = time.time() - start_time

print(f"Time on CPU: {cpu_time:.4f} seconds")

# move matrices to GPU
gpu1 = cpu1.to('cuda')
gpu2 = cpu2.to('cuda')

# measure time on cpu
start_time = time.time()
result_gpu = torch.matmul(gpu1,gpu2)
gpu_time = time.time() - start_time

print(f"Time on GPU: {gpu_time:.4f} seconds")
```

```
# Compare results
print("Speedup (cpu time / gpu time):", cpu_time // gpu_time)
```

### Common Uses of `unsqueeze()`

* **Adding Batch Dimensions:** Neural networks typically expect data in batches (e.g., `[batch_size, channels, height, width]`). Passing a single image with shape `[3, 64, 64]` requires `unsqueeze(0)` to become `[1, 3, 64, 64]`.
* **Broadcasting Compatibility:** Adjusting a 1D tensor to a 2D column or row vector allows correct matrix multiplication or element-wise math with other multi-dimensional tensors.
* **Channel Alignment:** Inserting color or feature dimensions at specific axes to satisfy layer requirements.

```
# unsqueeze
c = torch.rand(226,226,3)  # (226,226,3) is image matrix
c.unsqueeze(0).shape
```

### Common Uses of `squeeze()`

* **Removing Batch Dimensions:** After running a single item through a model, the output often retains a batch size of one (e.g., `[1, 10]`). Using `squeeze(0)` drops that extra axis to return a clean 1D array of predictions (`[10]`).
* **Preparing for Loss Functions:** Many loss functions and evaluation metrics expect a 1D target vector of labels (shape `[N]`). If your data loader outputs labels with an extra dimension (shape `[N, 1]`), `squeeze(1)` fixes it.
* **Saving/Plotting Images:** Image libraries like `matplotlib` cannot plot grayscale images if they have an explicit channel dimension of one (shape `[28, 28, 1]`). Calling `squeeze()` reduces it to `[28, 28]` for rendering.


```
# squeeze
d = torch.rand(1,10)
d.squeeze(0).shape
```
