# Import Libraries

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

# Load Data

```python
df = pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
df.head()
```

```python
df.shape
```

```python
df.drop(columns={"id", "Unnamed: 32"}, inplace=True)
```

```python
df.head()
```

```python
x = df.iloc[:,1:]
```

```python
y = df.iloc[:,0]
```

# Split Data

```python
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.2)
```

### Feature Scaling

```python
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.fit_transform(X_test)
```

```python
X_train
```

### Label Encoding

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.fit_transform(y_test)
```

```python
y_train
```

```python
print(type(X_train))
print(type(y_train))
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor.dtype
```

```python
X_train_tensor = X_train_tensor.to(torch.float32)
X_test_tensor = X_test_tensor.to(torch.float32)
y_train_tensor = y_train_tensor.to(torch.float32)
y_test_tensor = y_test_tensor.to(torch.float32)
```

### Why `float64` → `float32`?
* PyTorch's neural network layers (`nn.Linear`) use `float32` by default, while Pandas/NumPy often provide `float64`.
* Converting to `float32` avoids dtype mismatch errors and is faster and more memory-efficient.

# Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

# Loss Function 

Setting up the loss function (to measure error)

```python
# define loss function
loss_function = nn.BCELoss()
```

# Defining Model by PyTorch nn Module

```python
class MySimpleNN(nn.Module):

    def __init__(self,num_features):

      super().__init__() # call parent constructor
      self.linear = nn.Linear(num_features, 1)
      self.sigmoid = nn.Sigmoid()

    def forward(self, features):
      out = self.linear(features)
      out = self.sigmoid(out)

      return out
```

# Trainig PipeLine

* `y_pred` is `(455, 1)` because the Linear layer has 1 output neuron.
* `y_train_tensor` is `(455,)` because the original target is a 1D array.
* So we reshape it to `(455, 1)` to match `y_pred` before calculating the loss.

### Model Calling

Passing features to the model to get predictions (forward pass).

```python
# Step - 1 create model
model = MySimpleNN(X_train_tensor.shape[1])

# Step - 2 define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Step - 3 define loop for epoch
for epoch in range(epochs):

    # Step - 4 forward pass
    y_pred = model(X_train_tensor)

    # Step - 5 loss calculate
    loss = loss_function(y_pred,y_train_tensor.reshape(-1, 1))

    # Step - 6 Zero gradients
    optimizer.zero_grad()

    # Step - 7 backward pass
    loss.backward()

    # Step - 8 parameter update
    optimizer.step()

    print(f"epoch: {epoch + 1}, loss: {loss}")

```

```python
print(model.linear.weight)
print()
print(model.linear.bias)
```

# Evaluation

```python
# model evalution
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred = (y_pred >  0.9).float()
    accuracy = (y_pred == y_test_tensor).float().mean()
    print(accuracy)
```
